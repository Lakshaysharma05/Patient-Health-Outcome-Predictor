```python
X = df.drop("Condition_Improved", axis=1)
y = df["Condition_Improved"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
X = X.drop("Patient_ID", axis=1)

print("X shape:", X.shape)
```

```python
print(X.select_dtypes(include="object").columns)
```

```python
X = pd.get_dummies(X, columns=[
    "Gender",
    "Living_Area",
    "Past_Diseases",
    "Smoking_Status",
    "Diet_Quality",
    "Stress_Level",
    "Treatment",
    "Follow_Up"
], dtype=int)

print("X shape:", X.shape)
print(X.head())
```

```python
X_np = X.to_numpy()
y_np = y.to_numpy()

print("X_np shape:", X_np.shape)
print("y_np shape:", y_np.shape)
print("X_np type:", X_np.dtype)
print("y_np type:", y_np.dtype)
```

```python
np.random.seed(42)

indices = np.arange(len(X_np))
np.random.shuffle(indices)

split = int(0.8 * len(indices))

train_indices = indices[:split]
test_indices = indices[split:]

X_train = X_np[train_indices]
X_test = X_np[test_indices]

y_train = y_np[train_indices]
y_test = y_np[test_indices]

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
mean = X_train.mean(axis=0)
std = X_train.std(axis=0)

# Prevent division by zero
std[std == 0] = 1

X_train_scaled = (X_train - mean) / std
X_test_scaled = (X_test - mean) / std

print("Training data:")
print("Mean:", X_train_scaled.mean())
print("Std:", X_train_scaled.std())

print("\nShapes:")
print("X_train_scaled:", X_train_scaled.shape)
print("X_test_scaled:", X_test_scaled.shape)
```

```python
def sigmoid(z):
    z = np.clip(z, -500, 500)
    return 1 / (1 + np.exp(-z))

print(sigmoid(-2))
print(sigmoid(0))
print(sigmoid(2))
```

```python
weights = np.zeros(X_train_scaled.shape[1])
bias = 0.0

print("Weights shape:", weights.shape)
print("Bias:", bias)
```

```python
learning_rate = 0.01
epochs = 1000

for epoch in range(epochs):

    # 1. Calculate linear output
    z = X_train_scaled @ weights + bias

    # 2. Convert output to probability
    predictions = sigmoid(z)

    # 3. Calculate error
    error = predictions - y_train

    # 4. Calculate gradients
    dw = (X_train_scaled.T @ error) / len(X_train_scaled)
    db = np.mean(error)

    # 5. Update weights and bias
    weights -= learning_rate * dw
    bias -= learning_rate * db

print("Training completed!")
print("Final bias:", bias)
print("First 5 weights:", weights[:5])
```

```python
test_probabilities = sigmoid(X_test_scaled @ weights + bias)
y_pred = (test_probabilities >= 0.5).astype(int)

print(y_pred[:10])
```

```python
accuracy = np.mean(y_pred == y_test)

print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)
```

```python
tp = np.sum((y_pred == 1) & (y_test == 1))
tn = np.sum((y_pred == 0) & (y_test == 0))
fp = np.sum((y_pred == 1) & (y_test == 0))
fn = np.sum((y_pred == 0) & (y_test == 1))

print("True Positive (TP):", tp)
print("True Negative (TN):", tn)
print("False Positive (FP):", fp)
print("False Negative (FN):", fn)
```

```python
precision = tp / (tp + fp)
recall = tp / (tp + fn)
f1_score = 2 * (precision * recall) / (precision + recall)

print("Precision:", precision)
print("Precision (%):", precision * 100)

print("Recall:", recall)
print("Recall (%):", recall * 100)

print("F1-Score:", f1_score)
print("F1-Score (%):", f1_score * 100)
```
